```python
def predict_domain(text, model, tokenizer, label_encoder, max_length):
    model.eval() # Set model to evaluation mode
    inputs = tokenizer.encode_plus(
        text,
        None,
        add_special_tokens=True,
        max_length=max_length,
        pad_to_max_length=True,
        return_token_type_ids=True,
        truncation=True
    )
    ids = torch.tensor(inputs['input_ids'], dtype=torch.long).unsqueeze(0).to(device)
    mask = torch.tensor(inputs['attention_mask'], dtype=torch.long).unsqueeze(0).to(device)
    token_type_ids = torch.tensor(inputs['token_type_ids'], dtype=torch.long).unsqueeze(0).to(device)

    with torch.no_grad():
        outputs = model(ids, attention_mask=mask, token_type_ids=token_type_ids)
        logits = outputs.logits

    # Apply softmax to get probabilities and get the predicted class index
    probabilities = torch.softmax(logits, dim=1).cpu().numpy()
    predicted_class_idx = np.argmax(probabilities, axis=1)[0]

    # Decode the predicted class index to the original label
    predicted_label = label_encoder[predicted_class_idx]
    return predicted_label

# def predict_sub_domain(text, model, tokenizer, label_encoder, max_length):
#     model.eval() # Set model to evaluation mode
#     inputs = tokenizer.encode_plus(
#         text,
#         None,
#         add_special_tokens=True,
#         max_length=max_length,
#         pad_to_max_length=True,
#         return_token_type_ids=True,
#         truncation=True
#     )
#     ids = torch.tensor(inputs['input_ids'], dtype=torch.long).unsqueeze(0).to(device)
#     mask = torch.tensor(inputs['attention_mask'], dtype=torch.long).unsqueeze(0).to(device)
#     token_type_ids = torch.tensor(inputs['token_type_ids'], dtype=torch.long).unsqueeze(0).to(device)

#     with torch.no_grad():
#         outputs = model(ids, attention_mask=mask, token_type_ids=token_type_ids)
#         logits = outputs.logits

#     # Apply softmax to get probabilities and get the predicted class index
#     probabilities = torch.softmax(logits, dim=1).cpu().numpy()
#     predicted_class_idx = np.argmax(probabilities, axis=1)[0]

#     # Decode the predicted class index to the original label
#     predicted_label = label_encoder[predicted_class_idx]
#     return predicted_label

print("Prediction functions 'predict_domain' and 'predict_sub_domain' defined.")
```

## Prepare Data for Sub-Domain Classification

### Subtask:
Prepare the data for 'Sub-Domain' classification by creating a new LabelEncoder for the 'Sub-Domain' column, encoding it into numerical labels, and then splitting the `Text + Domain` input and encoded 'Sub-Domain' data into training, validation, and test sets, maintaining class distribution.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# Special token map for language neutrality (Fix from Fixes.txt)
domain_token_map = {0:"[SPORTS]", 1:"[CRIME]", 2:"[ENTMT]", 3:"[AGRI]", 4:"[BUSINESS]"}

sub_domain_label_encoder = LabelEncoder()
nlp_df['Sub-Domain_cleaned'] = nlp_df['Sub-Domain'].str.strip()

problematic_sub_domain_class = 'Sub-Domain'
if problematic_sub_domain_class in nlp_df['Sub-Domain_cleaned'].value_counts() and \
   nlp_df['Sub-Domain_cleaned'].value_counts()[problematic_sub_domain_class] == 1:
    nlp_df = nlp_df[nlp_df['Sub-Domain_cleaned'] != problematic_sub_domain_class].copy()

sub_domain_label_encoder.fit(nlp_df['Sub-Domain_cleaned'].unique())

# Apply the language-neutral prefix using IDs
X_sub_domain = nlp_df['Domain_ID'].map(domain_token_map) + " " + nlp_df['Text'].fillna('')
y_sub_domain_numerical = nlp_df['Sub-Domain_ID']

X_train_sub_domain, X_temp_sub_domain, y_train_sub_domain, y_temp_sub_domain = train_test_split(
    X_sub_domain, y_sub_domain_numerical, test_size=0.3, random_state=42, stratify=y_sub_domain_numerical
)

X_val_sub_domain, X_test_sub_domain, y_val_sub_domain, y_test_sub_domain = train_test_split(
    X_temp_sub_domain, y_temp_sub_domain, test_size=0.5, random_state=42, stratify=y_temp_sub_domain
)

print("Data prepared using language-neutral domain tokens.")
```

```python
import torch
from tqdm.auto import tqdm
from sklearn.model_selection import train_test_split

# 1. Prepare for batch inference to get predicted domains for the training set
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
loaded_domain_model.to(device)
loaded_domain_model.eval()

def get_all_predicted_domains(texts, model, tokenizer):
    predictions = []
    for text in tqdm(texts, desc="Predicting Domains for Sub-Domain training"):
        inputs = tokenizer(text, truncation=True, padding='max_length', max_length=256, return_tensors='pt').to(device)
        with torch.no_grad():
            logits = model(**inputs).logits
        pred_id = torch.argmax(logits, dim=-1).item()
        # Corrected: Using 'domain_label_encoder' instead of 'domain_id_to_name'
        predictions.append(domain_label_encoder.get(pred_id, 'Unknown'))
    return predictions

# 2. Add predicted domain labels to the dataframe
print("Generating domain features using the pre-trained domain model...")
nlp_df['Predicted_Domain_Feature'] = get_all_predicted_domains(nlp_df['Text'].fillna('').tolist(), loaded_domain_model, loaded_domain_tokenizer)

# 3. Create the new concatenated input
nlp_df['Text_With_Predicted_Domain'] = nlp_df['Predicted_Domain_Feature'] + " " + nlp_df['Text'].fillna('')

# 4. Re-split the data using these new features
X_train_sub, X_temp_sub, y_train_sub, y_temp_sub = train_test_split(
    nlp_df['Text_With_Predicted_Domain'],
    nlp_df['Sub-Domain_ID'],
    test_size=0.3,
    random_state=42,
    stratify=nlp_df['Sub-Domain_ID']
)

X_val_sub, X_test_sub, y_val_sub, y_test_sub = train_test_split(
    X_temp_sub, y_temp_sub, test_size=0.5, random_state=42, stratify=y_temp_sub
)

print("Data prepared using model-predicted domain features.")
```

### Updating the DatasetDict and Retraining
Now we update the tokenization step to use these new splits which incorporate the domain model's logic.

```python
from datasets import Dataset, DatasetDict
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer
import torch

# 1. Initialize Tokenizer
xlm_roberta_sub_domain_model_name = 'xlm-roberta-base'
xlm_roberta_sub_domain_tokenizer = AutoTokenizer.from_pretrained(xlm_roberta_sub_domain_model_name)

# 2. Define Tokenization Function
def tokenize_function_xlm_roberta_sub_domain(examples):
    tokenized_inputs = xlm_roberta_sub_domain_tokenizer(
        examples['text'],
        truncation=True,
        padding='max_length',
        max_length=512
    )
    tokenized_inputs['labels'] = examples['label']
    return tokenized_inputs

# 3. Prepare DatasetDict
dataset_dict_sub_domain_v2 = DatasetDict({
    'train': Dataset.from_dict({'text': X_train_sub.tolist(), 'label': y_train_sub.tolist()}),
    'validation': Dataset.from_dict({'text': X_val_sub.tolist(), 'label': y_val_sub.tolist()}),
    'test': Dataset.from_dict({'text': X_test_sub.tolist(), 'label': y_test_sub.tolist()})
})

# 4. Map and Process
processed_datasets_v2 = dataset_dict_sub_domain_v2.map(tokenize_function_xlm_roberta_sub_domain, batched=True)
processed_datasets_v2.set_format(type='torch', columns=['input_ids', 'attention_mask', 'labels'])

# 5. Initialize Model and Trainer to avoid NameError
num_labels = len(set(y_train_sub))
xlm_roberta_sub_domain_model = AutoModelForSequenceClassification.from_pretrained(
    xlm_roberta_sub_domain_model_name,
    num_labels=num_labels
)

training_args_v2 = TrainingArguments(
    output_dir='./results_v2',
    eval_strategy='epoch',
    learning_rate=2e-5,
    per_device_train_batch_size=8,
    num_train_epochs=3,
    weight_decay=0.01,
    report_to='none'
)

trainer_sub_domain = Trainer(
    model=xlm_roberta_sub_domain_model,
    args=training_args_v2,
    train_dataset=processed_datasets_v2['train'],
    eval_dataset=processed_datasets_v2['validation']
)

print("Trainer and datasets are now fully initialized and ready for retraining.")
```

## Handling Class Imbalance for Sub-Domain Classification

```python
import numpy as np
import torch
from sklearn.utils.class_weight import compute_class_weight

# Calculate class weights for sub-domain classification
classes_sub_domain = np.unique(y_train_sub_domain)
weights_sub_domain = compute_class_weight('balanced', classes=classes_sub_domain, y=y_train_sub_domain)
class_weights_sub_domain = torch.tensor(weights_sub_domain, dtype=torch.float)

print("Class weights for Sub-Domain classification calculated:")
print(class_weights_sub_domain)
```

```python
from torch.nn import CrossEntropyLoss
from transformers import Trainer

# Custom Trainer for weighted loss
class WeightedTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, num_items_in_batch=None):
        labels = inputs.pop('labels')
        outputs = model(**inputs)
        logits = outputs.logits
        loss_fn = CrossEntropyLoss(
            weight=class_weights_sub_domain.to(model.device)
        )
        loss = loss_fn(logits, labels)
        return (loss, outputs) if return_outputs else loss

print("WeightedTrainer class defined.")
```

##Update of Fine Tuning after 3 epochs(for better metrices)

```python
from transformers import AutoModelForSequenceClassification, AutoConfig, TrainingArguments, Trainer, EarlyStoppingCallback
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
import torch
from torch.nn import CrossEntropyLoss
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# ✅ FIX 3: Dropout বাড়াও — overfitting কমাবে
# ❌ prev: directly from_pretrained করা হতো, dropout default 0.1 ছিল
config = AutoConfig.from_pretrained(
    xlm_roberta_sub_domain_model_name,
    num_labels=nlp_df['Sub-Domain_ID'].nunique(),
    hidden_dropout_prob=0.2,
    attention_probs_dropout_prob=0.2
)
xlm_roberta_sub_domain_model = AutoModelForSequenceClassification.from_pretrained(
    xlm_roberta_sub_domain_model_name,
    config=config
)
print(f"Model loaded. Labels: {nlp_df['Sub-Domain_ID'].nunique()}")

# ✅ FIX 2: WeightedTrainer — Tennis imbalance fix
# ❌ prev: plain Trainer ছিল, minority class ignore হতো
classes = np.unique(y_train_sub_domain)
weights = compute_class_weight('balanced', classes=classes, y=y_train_sub_domain)
class_weights = torch.tensor(weights, dtype=torch.float)

class WeightedTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels = inputs.pop('labels')
        outputs = model(**inputs)
        loss_fn = CrossEntropyLoss(weight=class_weights.to(model.device))
        loss = loss_fn(outputs.logits, labels)
        return (loss, outputs) if return_outputs else loss

def compute_metrics(eval_pred):
    labels = eval_pred.label_ids
    preds  = eval_pred.predictions.argmax(-1)
    p, r, f1, _ = precision_recall_fscore_support(labels, preds, average='macro', zero_division=0)
    return {'accuracy': accuracy_score(labels, preds), 'f1': f1, 'precision': p, 'recall': r}

training_args_sub_domain = TrainingArguments(
    output_dir='./results_xlm_roberta_sub_domain',

    # ✅ FIX 1: 3→5 epoch — result দেখে বোঝা গেছে আরো শেখার সুযোগ আছে
    # ❌ prev: num_train_epochs=3 — accuracy এখনো বাড়ছিল, early stop করা ঠিক হয়নি
    num_train_epochs=10,

    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    gradient_accumulation_steps=2,
    learning_rate=2e-5,
    warmup_ratio=0.1,
    weight_decay=0.01,
    lr_scheduler_type='linear',
    logging_dir='./logs_xlm_roberta_sub_domain',
    logging_steps=10,
    eval_strategy='epoch',
    save_strategy='epoch',
    save_total_limit=1,
    load_best_model_at_end=True,
    metric_for_best_model='f1',
    greater_is_better=True,
    fp16=True,
    seed=42,
    report_to='none'
)

trainer_sub_domain = WeightedTrainer(
    model=xlm_roberta_sub_domain_model,
    args=training_args_sub_domain,
    train_dataset=processed_datasets_xlm_roberta_sub_domain['train'],
    eval_dataset=processed_datasets_xlm_roberta_sub_domain['validation'],
    compute_metrics=compute_metrics,
    # ✅ patience=3 — 3 epoch improve না হলে auto-stop, overfit হবে না
    callbacks=[EarlyStoppingCallback(early_stopping_patience=3)]
)

print("Trainer initialized.")
trainer_sub_domain.train()

results_xlm_roberta_sub_domain = trainer_sub_domain.evaluate(
    processed_datasets_xlm_roberta_sub_domain['test']
)

print("\nTest Results:")
for k, v in results_xlm_roberta_sub_domain.items():
    print(f"  {k}: {v:.4f}")

import pandas as pd
display(pd.DataFrame([{
    'Model': 'XLM-RoBERTa (Sub-Domain)',
    'Accuracy':  results_xlm_roberta_sub_domain['eval_accuracy'],
    'Precision': results_xlm_roberta_sub_domain['eval_precision'],
    'Recall':    results_xlm_roberta_sub_domain['eval_recall'],
    'F1-Score':  results_xlm_roberta_sub_domain['eval_f1']
}]))
```

```python
import plotly.graph_objects as go
import pandas as pd

def plot_interactive_loss(trainer, title):
    # Extract log history
    history = trainer.state.log_history

    train_loss = [entry for entry in history if 'loss' in entry and 'epoch' in entry]
    val_loss = [entry for entry in history if 'eval_loss' in entry and 'epoch' in entry]

    df_train = pd.DataFrame(train_loss)
    df_val = pd.DataFrame(val_loss)

    fig = go.Figure()

    # Add Training Loss Trace
    if not df_train.empty:
        fig.add_trace(go.Scatter(
            x=df_train['epoch'],
            y=df_train['loss'],
            mode='lines+markers',
            name='Training Loss',
            line=dict(color='blue')
        ))

    # Add Validation Loss Trace
    if not df_val.empty:
        fig.add_trace(go.Scatter(
            x=df_val['epoch'],
            y=df_val['eval_loss'],
            mode='lines+markers',
            name='Validation Loss',
            line=dict(color='red', dash='dash')
        ))

    fig.update_layout(
        title=f'Interactive Loss Curve: {title}',
        xaxis_title='Epochs',
        yaxis_title='Loss',
        template='plotly_white',
        hovermode='x unified'
    )

    fig.show()

# Generating Interactive Graph only for Sub-Domain Model
print("Generating Interactive Loss Graph for Sub-Domain Model...")
plot_interactive_loss(trainer_sub_domain, "Sub-Domain Classification (XLM-RoBERTa)")
```

```python
import torch
import numpy as np

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Domain to Sub-Domain ID mapping (Fix from Fixes.txt)
domain_to_subdomain_ids = {
    0: [0, 1, 2],    # Sports
    1: [3, 4, 5],    # Crime
    2: [6, 7, 8],    # Entertainment
    3: [9, 10, 11],  # Agriculture
    4: [12, 13, 14]  # Business
}

domain_token_map = {0:"[SPORTS]", 1:"[CRIME]", 2:"[ENTMT]", 3:"[AGRI]", 4:"[BUSINESS]"}

def predict_domain(text, model, tokenizer, max_length=256):
    inputs = tokenizer(text, truncation=True, padding='max_length', max_length=max_length, return_tensors='pt').to(device)
    with torch.no_grad():
        logits = model(**inputs).logits
    pred_id = torch.argmax(logits, dim=-1).item()
    return pred_id

def predict_sub_domain_masked(text, model, tokenizer, domain_id, max_length=512):
    # Prepend neutral token (Consistency fix)
    text_with_prefix = domain_token_map[domain_id] + " " + text
    inputs = tokenizer(text_with_prefix, truncation=True, padding='max_length', max_length=max_length, return_tensors='pt').to(device)

    with torch.no_grad():
        logits = model(**inputs).logits.squeeze(0)

    # Masking logic (Accuracy/Hallucination fix)
    valid_ids = domain_to_subdomain_ids[domain_id]
    mask = torch.full_like(logits, float('-inf'))
    mask[valid_ids] = logits[valid_ids]

    pred_id = torch.argmax(mask).item()
    return pred_id

print("Hierarchical masked prediction logic implemented.")
```
